# Análisis de Sentimiento con IMDb Reviews

# Objetivo

El objetivo de este trabajo es aplicar técnicas de análisis de sentimiento sobre un dataset de reseñas de películas de IMDb utilizando dos enfoques diferentes:

1. Un método basado en léxicos utilizando listas de palabras positivas y negativas.
2. Un modelo preentrenado de análisis de sentimiento utilizando TextBlob.

Además, se busca evaluar y comparar ambos métodos mediante métricas de clasificación como Accuracy, Precision, Recall y F1-score, con el fin de analizar sus ventajas, limitaciones y desempeño en tareas de procesamiento de lenguaje natural.

```python
import pandas as pd
from textblob import TextBlob

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

#Dataset
df = pd.read_csv(r"C:\Users\Carolina\Downloads\IMDB Dataset.csv")
df.head()

#Procesamiento
df["sentiment"] = df["sentiment"].map({
    "positive": 1,
    "negative": 0
})

#Test
X_train, X_test, y_train, y_test = train_test_split(
    df["review"],
    df["sentiment"],
    test_size=0.2,
    random_state=42
)

#Léxico
positive_words = [
    "good", "great", "excellent",
    "amazing", "love", "best"
]

negative_words = [
    "bad", "terrible", "worst",
    "awful", "hate", "boring"
]

#Función léxica
def lexical_sentiment(review):

    words = review.lower().split()

    positive_count = sum(
        word in positive_words
        for word in words
    )

    negative_count = sum(
        word in negative_words
        for word in words
    )

    if positive_count > negative_count:
        return 1
    else:
        return 0

y_pred_lexical = X_test.apply(lexical_sentiment) #Predicciones léxicas

#TextBlob
def textblob_sentiment(review):

    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        return 1
    else:
        return 0

y_pred_textblob = X_test.apply(textblob_sentiment)

#Métricas
def evaluate_model(y_true, y_pred, model_name):

    print(f"\n{model_name}")

    print("Accuracy:",
          accuracy_score(y_true, y_pred))

    print("Precision:",
          precision_score(y_true, y_pred))

    print("Recall:",
          recall_score(y_true, y_pred))

    print("F1-score:",
          f1_score(y_true, y_pred))

#Evaluación
evaluate_model(
    y_test,
    y_pred_lexical,
    "Enfoque Léxico"
)

evaluate_model(
    y_test,
    y_pred_textblob,
    "TextBlob"
)

```

# Comparación de resultados

El modelo TextBlob obtuvo mejores resultados generales que el enfoque léxico.

El enfoque léxico alcanzó un accuracy de 66.74%, mientras que TextBlob logró 69.26%.

Además, TextBlob obtuvo un F1-score superior, lo que indica un mejor equilibrio entre precision y recall.

Sin embargo, el enfoque léxico tuvo una precision ligeramente mayor, produciendo menos falsos positivos.
